```python

import pandas as pd
import numpy as np
# Phase 1: Data Preprocessing and Scaling

# 1. Load the CSV file into a Pandas DataFrame so we can read the housing data
train = pd.read_csv('train (2).csv')

# 2. Select our input feature (square footage) and our target label (house price), these are closely related to price
features = ['GrLivArea']
target = 'SalePrice'

# 3. Extract those columns as raw numbers and reshape the target into a vertical column vector
X_raw = train[features].values
y = train[target].values.reshape(-1, 1)

# 4. Calculate the mean (average) and standard deviation (spread) of our feature data
mu = np.mean(X_raw, axis=0)
sigma = np.std(X_raw, axis=0)

# 5. Standardize the feature data (Z-score normalization) to prevent math/overflow errors
X_scaled = (X_raw - mu) / sigma

# 6. Count how many total houses (rows) we have in our dataset
m = len(X_scaled)

# 7. Prepend a column of 1s to our matrix so gradient descent can calculate the y-intercept automatically
X = np.hstack([np.ones((m, 1)), X_scaled])

# 8. Print out the shapes to verify our matrices are properly formatted for linear algebra
print("Feature Matrix X shape:", X.shape)
print("Target Vector y shape:", y.shape)

# Phase 2: Using Mathematcal logic to predict and calculate error

def compute_hypothesis(X, theta):
    # Matrix multiplication: X (m x 2) dot theta (2 x 1) gives predictions (m x 1)
    return np.dot(X, theta)


def compute_cost(X, y, theta):
    m = len(y)
    predictions = compute_hypothesis(X, theta)

    # Squared error: (predictions - actual_y)^2
    squared_errors = (predictions - y) ** 2

    # Mean Squared Error formula
    cost = (1 / (2 * m)) * np.sum(squared_errors)
    return cost

# Phase 3: Simulating the code multiple times to make predictions, calculate error, readjust weights and then graphing the frop in error.

def gradient_descent(X, y, theta, alpha, iterations):
    m = len(y)
    cost_history = []

    for i in range(iterations):
        # 1. Make predictions using current weights
        predictions = compute_hypothesis(X, theta)

        # 2. Calculate the gradient (calculus derivative of MSE)
        gradients = (1 / m) * np.dot(X.T, (predictions - y))

        # 3. Update the weights by stepping downhill
        theta = theta - (alpha * gradients)

        # 4. Calculate and save the cost for this step
        cost = compute_cost(X, y, theta)
        cost_history.append(cost)

    return theta, cost_history


alpha = 0.01       # Learning rate (how big of a step to take)
iterations = 1000  # Number of training steps

# 2. Initialize starting weights (theta) to zeros
# X has 2 columns (intercept column + feature column), so theta needs 2 rows
theta = np.zeros((X.shape[1], 1))

# 3. Run your custom gradient descent engine
final_theta, cost_history = gradient_descent(X, y, theta, alpha, iterations)

# 4. Print out your results
print("Final Optimized Weights (theta):")
print(final_theta)
print("\nFinal Cost (MSE):", cost_history[-1])

# 5. Plot the convergence using matplotlib to watch the error slide down[cite: 3]
import matplotlib.pyplot as plt

plt.plot(range(iterations), cost_history, color='blue', linewidth=2)
plt.xlabel('Iterations')
plt.ylabel('Cost (MSE)')
plt.title('Gradient Descent Convergence')
plt.grid(True)
plt.show()
```
